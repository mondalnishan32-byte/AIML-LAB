```python
from collections import deque
capacity_A = int(input("Enter capacity of Jug A: "))
capacity_B = int(input("Enter capacity of Jug B: "))
target = int(input("Enter target amount: "))

def water_jug_bfs():
    start = (0, 0)
    queue = deque()
    queue.append((start, []))
    visited = set()
    visited.add(start)
    while queue:
        (a, b), path = queue.popleft()
        if a == target or b == target:
            return path + [(a, b)]
        states = []
        states.append(((capacity_A, b), "Fill Jug A"))

        states.append(((a, capacity_B), "Fill Jug B"))

        states.append(((0, b), "Empty Jug A"))

        states.append(((a, 0), "Empty Jug B"))

        amount = min(a, capacity_B - b)
        states.append((
            (a - amount, b + amount),
            "Pour Jug A into Jug B"
        ))
        amount = min(b, capacity_A - a)
        states.append((
            (a + amount, b - amount),
            "Pour Jug B into Jug A"
        ))
        for new_state, operation in states:
            if new_state not in visited:
                visited.add(new_state)
                queue.append(
                    (new_state, path + [(operation, new_state)])
                )
    return None
solution = water_jug_bfs()
if solution:
    print("\nMinimum sequence of steps:")
    print("Step 0: (0, 0)")

    for i, (operation, state) in enumerate(solution[:-1], 1):
        print(f"Step {i}: {operation} -> {state}")

    print(f"Step {len(solution)-1}: Goal reached -> {solution[-1]}")
else:
    print("\nNo solution exists.")
```
